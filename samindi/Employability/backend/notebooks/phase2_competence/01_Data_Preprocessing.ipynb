# Phase 02 – Competence Data Preprocessing

## Objective

The purpose of this notebook is to prepare the Phase 02 competence dataset for statistical analysis and machine learning.

This phase investigates the relationship between **Competence** and **Employability**.

- Predictor construct: Competence
- Competence items: C1–C10
- Target construct: Employability
- Employability items: EVS1–EVS12
- Measurement scale: 5-point Likert scale

The preprocessing stage includes data loading, structure validation, missing-value checks, duplicate checks, Likert-scale validation, composite-score creation, and preparation of the cleaned dataset for the next stages of the research.

## 1. Import Required Libraries

The required Python libraries are imported for file-path handling, data manipulation, and numerical processing.

```python
from pathlib import Path
import pandas as pd
import numpy as np
```

## 2. Define Project Paths

The project directory and input/output file paths are defined to maintain a consistent and reproducible project structure.

The raw Phase 02 dataset is loaded from the `data/raw` directory, while the prepared dataset will be saved in the `data/processed` directory.

```python
PROJECT_DIR = Path.cwd()

if PROJECT_DIR.name == "phase2_competence":
    PROJECT_DIR = PROJECT_DIR.parent.parent

CANDIDATE_RAW_FILES = [
    PROJECT_DIR / "data" / "raw" / "phase02_competence_new.csv",
    PROJECT_DIR / "data" / "raw" / "phase2_competence_dataset_final.csv",
]

RAW_DATA_FILE = next(
    (path for path in CANDIDATE_RAW_FILES if path.exists()),
    CANDIDATE_RAW_FILES[0],
)

PROCESSED_DIR = (
    PROJECT_DIR
    / "data"
    / "processed"
)

PROCESSED_DIR.mkdir(
    parents=True,
    exist_ok=True
)

OUTPUT_FILE = (
    PROCESSED_DIR
    / "phase2_competence_prepared.csv"
)

print("Project directory:", PROJECT_DIR)
print("Raw dataset:", RAW_DATA_FILE)
print("Processed output:", OUTPUT_FILE)
```

```python
if not RAW_DATA_FILE.exists():
    raise FileNotFoundError(
        f"Dataset not found: {RAW_DATA_FILE}"
    )

print("Raw Phase 02 dataset found successfully.")
```

## 3. Load the Raw Phase 02 Dataset

The raw questionnaire dataset is loaded into a Pandas DataFrame.

At this stage, the dataset is inspected without modifying the original data.

```python
df = pd.read_csv(RAW_DATA_FILE)

print("Dataset shape:", df.shape)

df.head()
```

```python
print("Dataset columns:")

for i, col in enumerate(df.columns, start=1):
    print(f"{i:02d}. {col}")
```

Define research variables

```python
competence_cols = [
    "C1", "C2", "C3", "C4", "C5",
    "C6", "C7", "C8", "C9", "C10"
]

evs_cols = [
    "EVS1", "EVS2", "EVS3", "EVS4",
    "EVS5", "EVS6", "EVS7", "EVS8",
    "EVS9", "EVS10", "EVS11", "EVS12"
]

tracking_cols = [
    "Response_ID",
    "data_source"
]

print("Competence items:", competence_cols)
print("Number of competence items:", len(competence_cols))

print("\nEmployability items:", evs_cols)
print("Number of employability items:", len(evs_cols))
```

## 4. Validate the Dataset Structure

The dataset column names are checked to confirm that all required Competence and Employability questionnaire items are available.

The expected research variables are:

- Competence: C1–C10
- Employability: EVS1–EVS12
- Tracking variables: Response_ID and data_source

Verify required columns exist

```python
required_cols = (
    tracking_cols
    + competence_cols
    + evs_cols
)

missing_cols = [
    col
    for col in required_cols
    if col not in df.columns
]

if missing_cols:
    raise ValueError(
        f"Missing required columns: {missing_cols}"
    )

print("All required Phase 02 columns are available.")
```

```python
df.info()
```

Check missing values

## 5. Missing-Value Analysis

Missing values are checked before performing any statistical analysis or model training.

This step is important because incomplete questionnaire responses may affect composite scores, reliability analysis, and machine-learning performance.

```python
missing_summary = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (
        df.isnull().sum()
        / len(df)
        * 100
    )
})

missing_summary
```

```python
total_missing = df.isnull().sum().sum()

print(
    "Total missing values:",
    total_missing
)
```

## 6. Duplicate and Response-ID Validation

Duplicate response identifiers and completely duplicated rows are examined.

Identical questionnaire responses are not automatically removed because different respondents may legitimately provide the same Likert-scale answers. Therefore, duplicate removal is performed only if there is clear evidence of duplicated records.

Check duplicate Response IDs

```python
df.duplicated()
```

```python
duplicate_ids = df[
    "Response_ID"
].duplicated().sum()

print(
    "Duplicate Response_ID values:",
    duplicate_ids
)
```

```python
full_duplicates = df.duplicated().sum()

print(
    "Completely duplicated rows:",
    full_duplicates
)
```

Check the source composition

```python
source_counts = df[
    "data_source"
].value_counts(
    dropna=False
)

print(source_counts)
```

```python
source_percentages = (
    df["data_source"]
    .value_counts(
        normalize=True
    )
    .mul(100)
    .round(2)
)

print(source_percentages)
```

Convert questionnaire values to numeric

```python
questionnaire_cols = (
    competence_cols
    + evs_cols
)

for col in questionnaire_cols:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )
```

```python
conversion_missing = (
    df[questionnaire_cols]
    .isnull()
    .sum()
)

print(
    conversion_missing[
        conversion_missing > 0
    ]
)
```

## 7. Data Source Verification

The dataset contains both original and TVAE-generated synthetic observations.

The `data_source` variable is retained only for provenance and validation purposes. It will not be used as a predictor in the employability model.

This separation is important to avoid introducing data-source information into the predictive model.

Validate Likert scale

```python
invalid_values = {}

for col in questionnaire_cols:

    invalid = df.loc[
        ~df[col].between(1, 5),
        col
    ].dropna()

    if not invalid.empty:
        invalid_values[col] = (
            sorted(
                invalid.unique()
            )
        )

if invalid_values:
    print(
        "Invalid Likert values found:"
    )

    for col, values in invalid_values.items():
        print(
            col,
            values
        )
else:
    print(
        "All questionnaire responses are within the valid 1–5 Likert range."
    )
```

Check unique values

```python
for col in questionnaire_cols:
    print(
        col,
        sorted(
            df[col]
            .dropna()
            .unique()
        )
    )
```

Create clean working dataframe

```python
phase2_df = df[
    [
        "Response_ID"
    ]
    + competence_cols
    + evs_cols
    + [
        "data_source"
    ]
].copy()

print(
    "Working dataset shape:",
    phase2_df.shape
)

phase2_df.head()
```

```python
phase2_df[
    "Comp_avg"
] = phase2_df[
    competence_cols
].mean(
    axis=1
)
```

```python
phase2_df[
    competence_cols
    + ["Comp_avg"]
].head()
```

```python
phase2_df[
    "EVS_avg"
] = phase2_df[
    evs_cols
].mean(
    axis=1
)
```

```python
phase2_df[
    evs_cols
    + ["EVS_avg"]
].head()
```

Create validated Competence subdimension scores

```python
life_career_cols = [
    "C1", "C2", "C3", "C4"
]

learning_innovation_cols = [
    "C5", "C6", "C7"
]

it_media_cols = [
    "C8", "C9", "C10"
]
```

```python
phase2_df[
    "LifeCareer_avg"
] = phase2_df[
    life_career_cols
].mean(axis=1)

phase2_df[
    "LearningInnovation_avg"
] = phase2_df[
    learning_innovation_cols
].mean(axis=1)

phase2_df[
    "ITMedia_avg"
] = phase2_df[
    it_media_cols
].mean(axis=1)
```

```python
human_capital_cols = [
    "EVS1", "EVS2", "EVS3"
]

social_capital_cols = [
    "EVS4", "EVS5"
]

career_identity_cols = [
    "EVS6", "EVS7",
    "EVS8", "EVS9"
]

environmental_monitoring_cols = [
    "EVS10"
]

balance_cols = [
    "EVS11", "EVS12"
]
```

```python
phase2_df[
    "HumanCapital_avg"
] = phase2_df[
    human_capital_cols
].mean(axis=1)

phase2_df[
    "SocialCapital_avg"
] = phase2_df[
    social_capital_cols
].mean(axis=1)

phase2_df[
    "CareerIdentity_avg"
] = phase2_df[
    career_identity_cols
].mean(axis=1)

phase2_df[
    "EnvironmentalMonitoring_avg"
] = phase2_df[
    environmental_monitoring_cols
].mean(axis=1)

phase2_df[
    "Balance_avg"
] = phase2_df[
    balance_cols
].mean(axis=1)
```

Validate calculated scores

```python
composite_cols = [
    "Comp_avg",
    "LifeCareer_avg",
    "LearningInnovation_avg",
    "ITMedia_avg",
    "EVS_avg",
    "HumanCapital_avg",
    "SocialCapital_avg",
    "CareerIdentity_avg",
    "EnvironmentalMonitoring_avg",
    "Balance_avg"
]

phase2_df[
    composite_cols
].describe().T
```

Final data-quality check

```python
print(
    "Final rows:",
    len(phase2_df)
)

print(
    "Final columns:",
    phase2_df.shape[1]
)

print(
    "Missing values:",
    phase2_df.isnull()
    .sum()
    .sum()
)

print(
    "Comp_avg range:",
    phase2_df[
        "Comp_avg"
    ].min(),
    "to",
    phase2_df[
        "Comp_avg"
    ].max()
)

print(
    "EVS_avg range:",
    phase2_df[
        "EVS_avg"
    ].min(),
    "to",
    phase2_df[
        "EVS_avg"
    ].max()
)
```

Save processed dataset

```python
phase2_df.to_csv(
    OUTPUT_FILE,
    index=False
)

print(
    "Prepared Phase 02 dataset saved successfully."
)

print(
    "File:",
    OUTPUT_FILE
)
```

Reload and verify saved dataset

```python
prepared_check = pd.read_csv(
    OUTPUT_FILE
)

print(
    "Reloaded shape:",
    prepared_check.shape
)

prepared_check.head()
```
