```python
competence_cols = [
    "C1", "C2", "C3", "C4", "C5",
    "C6", "C7", "C8", "C9", "C10"
]

life_career_cols = [
    "C1", "C2", "C3", "C4"
]

learning_innovation_cols = [
    "C5", "C6", "C7"
]

it_media_cols = [
    "C8", "C9", "C10"
]


evs_cols = [
    "EVS1", "EVS2", "EVS3", "EVS4",
    "EVS5", "EVS6", "EVS7", "EVS8",
    "EVS9", "EVS10", "EVS11", "EVS12"
]

human_capital_cols = [
    "EVS1", "EVS2", "EVS3"
]

social_capital_cols = [
    "EVS4", "EVS5"
]

career_identity_cols = [
    "EVS6", "EVS7", "EVS8", "EVS9"
]

balance_cols = [
    "EVS11", "EVS12"
]
```

## 2. Cronbach's Alpha

Cronbach's Alpha measures the internal consistency of a set of questionnaire items.

A higher Alpha generally indicates that the items measure the same underlying construct more consistently.

Common interpretation guidelines are:

- α ≥ 0.90 — Excellent
- 0.80 ≤ α < 0.90 — Good
- 0.70 ≤ α < 0.80 — Acceptable
- 0.60 ≤ α < 0.70 — Questionable
- 0.50 ≤ α < 0.60 — Poor
- α < 0.50 — Unacceptable

These thresholds are used as general guidance and should be interpreted together with the theoretical structure of the questionnaire.

```python
def cronbach_alpha(items_df):

    items_df = items_df.dropna()

    k = items_df.shape[1]

    if k < 2:
        return np.nan

    item_variances = items_df.var(
        axis=0,
        ddof=1
    )

    total_score = items_df.sum(axis=1)

    total_variance = total_score.var(
        ddof=1
    )

    if total_variance == 0:
        return np.nan

    alpha = (
        k / (k - 1)
    ) * (
        1 - (
            item_variances.sum()
            / total_variance
        )
    )

    return alpha
```

Calculate reliability

```python
reliability_results = {
    "Overall Competence (C1-C10)":
        cronbach_alpha(
            df[competence_cols]
        ),

    "Life and Career (C1-C4)":
        cronbach_alpha(
            df[life_career_cols]
        ),

    "Learning and Innovation (C5-C7)":
        cronbach_alpha(
            df[learning_innovation_cols]
        ),

    "IT and Media (C8-C10)":
        cronbach_alpha(
            df[it_media_cols]
        ),

    "Overall Employability (EVS1-EVS12)":
        cronbach_alpha(
            df[evs_cols]
        ),

    "Human Capital (EVS1-EVS3)":
        cronbach_alpha(
            df[human_capital_cols]
        ),

    "Social Capital (EVS4-EVS5)":
        cronbach_alpha(
            df[social_capital_cols]
        ),

    "Career Identity (EVS6-EVS9)":
        cronbach_alpha(
            df[career_identity_cols]
        ),

    "Balance (EVS11-EVS12)":
        cronbach_alpha(
            df[balance_cols]
        )
}
```

```python
reliability_df = pd.DataFrame(
    reliability_results.items(),
    columns=[
        "Construct",
        "Cronbach_Alpha"
    ]
)

reliability_df[
    "Cronbach_Alpha"
] = reliability_df[
    "Cronbach_Alpha"
].round(4)

reliability_df
```

## 3. Corrected Item–Total Correlation

Corrected item–total correlation measures the relationship between each questionnaire item and the total score calculated from the remaining items.

This analysis helps identify items that may not behave consistently with the overall construct.

Items are not automatically removed based only on statistical results. Any decision to remove an item must also consider the validated questionnaire structure and theoretical justification.

Item-total correlation function

```python
def corrected_item_total_correlations(items_df):

    results = {}

    for col in items_df.columns:

        other_cols = [
            c for c in items_df.columns
            if c != col
        ]

        other_total = (
            items_df[other_cols]
            .sum(axis=1)
        )

        correlation = (
            items_df[col]
            .corr(other_total)
        )

        results[col] = correlation

    return pd.Series(
        results,
        name="Corrected_Item_Total_Correlation"
    )
```

```python
competence_item_total = (
    corrected_item_total_correlations(
        df[competence_cols]
    )
)

competence_item_total
```

```python
evs_item_total = (
    corrected_item_total_correlations(
        df[evs_cols]
    )
)

evs_item_total
```

Alpha if item deleted

```python
def alpha_if_item_deleted(items_df):

    results = {}

    for col in items_df.columns:

        remaining_items = (
            items_df.drop(
                columns=[col]
            )
        )

        results[col] = (
            cronbach_alpha(
                remaining_items
            )
        )

    return pd.Series(
        results,
        name="Alpha_If_Item_Deleted"
    )
```

```python
competence_alpha_deleted = (
    alpha_if_item_deleted(
        df[competence_cols]
    )
)

competence_alpha_deleted
```

```python
evs_alpha_deleted = (
    alpha_if_item_deleted(
        df[evs_cols]
    )
)

evs_alpha_deleted
```

Combine item diagnostics

```python
competence_diagnostics = pd.concat(
    [
        competence_item_total,
        competence_alpha_deleted
    ],
    axis=1
)

competence_diagnostics.round(4)
```

```python
evs_diagnostics = pd.concat(
    [
        evs_item_total,
        evs_alpha_deleted
    ],
    axis=1
)

evs_diagnostics.round(4)
```
