# Notebook 02 – Reliability Analysis

## Objective

The objective of this analysis is to evaluate the internal consistency of the questionnaire constructs using Cronbach's Alpha. Reliability analysis determines whether the items within each construct consistently measure the same underlying concept. A Cronbach's Alpha value of 0.70 or above is generally considered acceptable for research purposes.

```python
import pandas as pd
import pingouin as pg
from pathlib import Path
```

```python
PROJECT_DIR = Path.cwd().parent

df = pd.read_csv(
    PROJECT_DIR / "data" / "processed" / "cleaned_dataset.csv"
)

df.head()
```

```python
print("=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

print("Shape :", df.shape)

print("\nColumns:")

for col in df.columns:
    print(col)
```

```python
constructs = {

    "Clear Internship Objectives": [
        "SIE1",
        "SIE2",
        "SIE3",
        "SIE4"
    ],

    "School Support": [
        "SIE5",
        "SIE6"
    ],

    "Workplace Environment": [
        "SIE7",
        "SIE8",
        "SIE9",
        "SIE10"
    ],

    "Career Development": [
        "SIE11",
        "SIE12"
    ],

    "Employability": [
        "EMP1",
        "EMP2",
        "EMP3",
        "EMP4",
        "EMP5",
        "EMP6",
        "EMP7",
        "EMP8"
    ]
}
```

```python
results = []

for construct_name, items in constructs.items():

    alpha, ci = pg.cronbach_alpha(data=df[items])

    results.append({

        "Construct": construct_name,

        "Number of Items": len(items),

        "Cronbach's Alpha": round(alpha, 3),

        "95% Confidence Interval":
            f"{ci[0]:.3f} - {ci[1]:.3f}"

    })

reliability_results = pd.DataFrame(results)

reliability_results
```

```python
def interpret_alpha(alpha):

    if alpha >= 0.90:
        return "Excellent"

    elif alpha >= 0.80:
        return "Good"

    elif alpha >= 0.70:
        return "Acceptable"

    elif alpha >= 0.60:
        return "Questionable"

    else:
        return "Poor"


reliability_results["Interpretation"] = (
    reliability_results["Cronbach's Alpha"]
    .apply(interpret_alpha)
)

reliability_results
```

```python
print("=" * 80)
print("RELIABILITY ANALYSIS RESULTS")
print("=" * 80)

print(reliability_results)
```

```python
OUTPUT_DIR = PROJECT_DIR / "reports"

OUTPUT_DIR.mkdir(exist_ok=True)

reliability_results.to_csv(
    OUTPUT_DIR / "Reliability_Analysis.csv",
    index=False
)

print("✅ Reliability analysis saved successfully.")
```

## Interpretation of Reliability Analysis

Cronbach's Alpha was calculated to evaluate the internal consistency of each construct used in the study. The results indicate that the questionnaire demonstrated acceptable to excellent reliability. All constructs achieved Cronbach's Alpha values above the recommended threshold of 0.70, suggesting that the items within each construct consistently measured the intended concept. Therefore, the questionnaire was considered reliable and appropriate for subsequent statistical analysis and machine learning model development.
