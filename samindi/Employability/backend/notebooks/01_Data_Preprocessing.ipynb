# 3. Missing Value Analysis

The purpose of this section is to identify missing values within the dataset before conducting statistical analysis and machine learning. Missing data can affect the reliability of research findings and model performance. Therefore, the dataset is examined to determine the number and percentage of missing values for each variable, enabling appropriate preprocessing decisions.

```python
print("=" * 70)
print("MISSING VALUE COUNT")
print("=" * 70)

missing_count = df.isnull().sum()

print(missing_count)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage.round(2)
})

missing_summary
```

```python
missing_only = missing_summary[
    missing_summary["Missing Count"] > 0
]

missing_only
```

```python
import matplotlib.pyplot as plt

missing_only.plot(
    kind="bar",
    y="Missing Count",
    figsize=(10,5)
)

plt.title("Missing Values by Variable")
plt.xlabel("Variables")
plt.ylabel("Missing Count")
plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```
