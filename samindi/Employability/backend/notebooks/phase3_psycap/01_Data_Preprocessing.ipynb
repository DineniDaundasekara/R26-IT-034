# Phase 03 — Psychological Capital → Employability

## Step 01 — Data Preprocessing

This notebook prepares the Phase 03 dataset for statistical analysis and machine learning.

The preprocessing stage includes:

- Loading the raw Phase 03 dataset
- Validating required columns
- Checking missing values and duplicate responses
- Validating 5-point Likert-scale responses
- Creating Psychological Capital construct scores
- Creating Employability construct scores
- Creating overall PsyCap and Employability averages
- Inspecting demographic variables
- Saving the cleaned and prepared dataset for later analysis

No predictive model is trained in this step.

```python
from pathlib import Path

import numpy as np
import pandas as pd
```

```python
from pathlib import Path

PROJECT_DIR = Path.cwd()

if PROJECT_DIR.name == "phase3_psycap":
    PROJECT_DIR = PROJECT_DIR.parent.parent

CANDIDATE_RAW_FILES = [
    PROJECT_DIR / "data" / "raw" / "phase03_psycap_new.csv",
    PROJECT_DIR / "data" / "raw" / "phase3_psycap_raw.csv",
]

RAW_DATA_PATH = next(
    (path for path in CANDIDATE_RAW_FILES if path.exists()),
    CANDIDATE_RAW_FILES[0],
)

PROCESSED_DIR = (
    PROJECT_DIR
    / "data"
    / "processed"
)

REPORT_DIR = (
    PROJECT_DIR
    / "reports"
    / "phase3"
    / "preprocessing"
)

PROCESSED_DIR.mkdir(
    parents=True,
    exist_ok=True
)

REPORT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print("PROJECT_DIR:", PROJECT_DIR)
print("RAW_DATA_PATH:", RAW_DATA_PATH)
print("PROCESSED_DIR:", PROCESSED_DIR)
print("REPORT_DIR:", REPORT_DIR)
```

```python
df = pd.read_csv(
    RAW_DATA_PATH
)

print(
    "Dataset shape:",
    df.shape
)

df.head()
```

Inspect All Columns

```python
print(
    "Number of columns:",
    len(df.columns)
)

print(
    "\nColumn names:"
)

for column in df.columns:
    print(column)
```

```python
COLUMN_RENAME_MAP = {

    # Identification
    "Response ID": "Response_ID",

    # Psychological Capital — Hope
    "I can find various ways to overcome obstacles that arise during my studies.": "PsyCap1",

    "I realize that different approaches can be used to overcome the problems I face in my studies.": "PsyCap2",

    "I am committed to doing my best and working hard to achieve the goals I set.": "PsyCap3",

    # Psychological Capital — Optimism
    "I have high hopes that I can succeed in my learning activities.": "PsyCap4",

    "I am optimistic that the time and effort I invest now will bring positive results in the future.": "PsyCap5",

    "I tend to identify positive lessons from my learning experiences, whether they involve success or failure.": "PsyCap6",

    # Psychological Capital — Resilience
    "I can identify and analyze problems that arise during my learning activities.": "PsyCap7",

    "When I receive low grades, I try to identify the reasons and find ways to improve.": "PsyCap8",

    "I understand that failure can happen, and I am able to recover and continue working toward long-term success.": "PsyCap9",

    # Psychological Capital — Self-Efficacy
    "I believe that I can successfully complete difficult tasks.": "PsyCap10",

    "I am confident enough to ask my teacher or lecturer for help when I have difficulty completing an assignment.": "PsyCap11",

    "I am confident that I can make a meaningful and useful contribution to others using my abilities.": "PsyCap12",

    # Employability
    "I have the ability to discuss professionally with various professionals or technical aspects of work": "EVS1",

    "I achieved a level of education sufficient to operate as a prospective new employee in the job": "EVS2",

    "I am able to evaluate sufficient technical skills to be able to complete tasks or solve problems at school": "EVS3",

    "I have a satisfactory level of professional experience to operate on the job": "EVS4",

    "I am able to convince others about work when decisions within the team have to be made": "EVS5",

    "Building social relationships can have a positive effect on my life": "EVS6",

    "I have a clear plan for my career": "EVS7",

    "I have clear career goals": "EVS8",

    "I have a clear strategy to achieve my career goals": "EVS9",

    "I have a clear idea of the target industry according to my field of work": "EVS10",

    "I am able to achieve a balance between my work goals and the support of my colleagues": "EVS11",

    "I am able to achieve a balance between my career goals and the support of my colleagues": "EVS12",

    # Data source
    "Data Source": "data_source"
}
```

```python
df = df.rename(
    columns=COLUMN_RENAME_MAP
)

print(
    "Columns renamed successfully."
)

print(
    "Number of columns:",
    len(df.columns)
)

for column in df.columns:
    print(column)
```

Define Column Groups

```python
PSYCAP_ITEMS = [
    "PsyCap1",
    "PsyCap2",
    "PsyCap3",
    "PsyCap4",
    "PsyCap5",
    "PsyCap6",
    "PsyCap7",
    "PsyCap8",
    "PsyCap9",
    "PsyCap10",
    "PsyCap11",
    "PsyCap12"
]

EVS_ITEMS = [
    "EVS1",
    "EVS2",
    "EVS3",
    "EVS4",
    "EVS5",
    "EVS6",
    "EVS7",
    "EVS8",
    "EVS9",
    "EVS10",
    "EVS11",
    "EVS12"
]

DEMOGRAPHIC_COLUMNS = [
    "Gender",
    "Age",
    "Degree Program",
    "Internship Status"
]
```

Define PsyCap Construct Mapping

```python
PSYCAP_CONSTRUCTS = {

    "Hope_avg": [
        "PsyCap1",
        "PsyCap2",
        "PsyCap3"
    ],

    "Optimism_avg": [
        "PsyCap4",
        "PsyCap5",
        "PsyCap6"
    ],

    "Resilience_avg": [
        "PsyCap7",
        "PsyCap8",
        "PsyCap9"
    ],

    "SelfEfficacy_avg": [
        "PsyCap10",
        "PsyCap11",
        "PsyCap12"
    ]
}
```

Define EVS Construct Mapping

```python
EVS_CONSTRUCTS = {

    "HumanCapital_avg": [
        "EVS1",
        "EVS2",
        "EVS3"
    ],

    "SocialCapital_avg": [
        "EVS4",
        "EVS5"
    ],

    "CareerIdentity_avg": [
        "EVS6",
        "EVS7",
        "EVS8",
        "EVS9"
    ],

    "EnvironmentalMonitoring_avg": [
        "EVS10"
    ],

    "Balance_avg": [
        "EVS11",
        "EVS12"
    ]
}
```

Validate Required Columns

```python
required_columns = (
    ["Response_ID"]
    + DEMOGRAPHIC_COLUMNS
    + PSYCAP_ITEMS
    + EVS_ITEMS
)

missing_columns = [
    column
    for column in required_columns
    if column not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

print(
    "Required column validation passed."
)
```

Check Duplicate Response IDs

```python
duplicate_response_ids = (
    df["Response_ID"]
    .duplicated()
    .sum()
)

print(
    "Duplicate Response_ID count:",
    duplicate_response_ids
)
```

```python
duplicate_rows = df[
    df["Response_ID"].duplicated(
        keep=False
    )
]

duplicate_rows
```

Check Exact Duplicate Rows

```python
exact_duplicate_rows = (
    df.duplicated()
    .sum()
)

print(
    "Exact duplicate row count:",
    exact_duplicate_rows
)
```

Missing Value Check

```python
missing_summary = (
    df[required_columns]
    .isnull()
    .sum()
    .sort_values(
        ascending=False
    )
)

missing_summary
```

```python
total_missing = int(
    missing_summary.sum()
)

print(
    "Total missing values:",
    total_missing
)
```

Validate Numeric Item Types

```python
for column in (
    PSYCAP_ITEMS
    +
    EVS_ITEMS
):

    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )
```

```python
numeric_conversion_missing = (
    df[
        PSYCAP_ITEMS
        +
        EVS_ITEMS
    ]
    .isnull()
    .sum()
)

numeric_conversion_missing
```

Validate Likert Range

```python
likert_columns = (
    PSYCAP_ITEMS
    +
    EVS_ITEMS
)

invalid_likert_values = {}

for column in likert_columns:

    invalid_count = (
        ~df[column].between(
            1,
            5
        )
    ).sum()

    if invalid_count > 0:
        invalid_likert_values[
            column
        ] = int(
            invalid_count
        )

print(
    "Invalid Likert values:",
    invalid_likert_values
)
```

Check Item Min/Max

```python
item_range_summary = pd.DataFrame({

    "Minimum": (
        df[likert_columns]
        .min()
    ),

    "Maximum": (
        df[likert_columns]
        .max()
    )
})

item_range_summary
```

Create PsyCap Construct Scores

```python
for construct_name, item_list in (
    PSYCAP_CONSTRUCTS.items()
):

    df[construct_name] = (
        df[item_list]
        .mean(
            axis=1
        )
    )
```

Create Overall PsyCap Score

```python
df["PsyCap_avg"] = (
    df[PSYCAP_ITEMS]
    .mean(
        axis=1
    )
)
```

Create EVS Construct Scores

```python
for construct_name, item_list in (
    EVS_CONSTRUCTS.items()
):

    df[construct_name] = (
        df[item_list]
        .mean(
            axis=1
        )
    )
```

Create Overall Employability Score

```python
df["EVS_avg"] = (
    df[EVS_ITEMS]
    .mean(
        axis=1
    )
)
```

Verify Construct Calculations

```python
construct_columns = [
    "Hope_avg",
    "Optimism_avg",
    "Resilience_avg",
    "SelfEfficacy_avg",
    "PsyCap_avg",

    "HumanCapital_avg",
    "SocialCapital_avg",
    "CareerIdentity_avg",
    "EnvironmentalMonitoring_avg",
    "Balance_avg",
    "EVS_avg"
]

df[
    construct_columns
].head()
```

Validate Construct Ranges

```python
construct_range_summary = pd.DataFrame({

    "Minimum": (
        df[
            construct_columns
        ]
        .min()
    ),

    "Maximum": (
        df[
            construct_columns
        ]
        .max()
    )
})

construct_range_summary
```

Automatic Range Validation

```python
invalid_constructs = {}

for column in construct_columns:

    invalid_count = (
        ~df[column].between(
            1,
            5
        )
    ).sum()

    if invalid_count > 0:

        invalid_constructs[
            column
        ] = int(
            invalid_count
        )

print(
    "Invalid construct scores:",
    invalid_constructs
)
```

Manually Verify One Respondent

```python
sample_index = 0

sample_psycap = df.loc[
    sample_index,
    PSYCAP_ITEMS
]

manual_psycap_average = (
    sample_psycap.mean()
)

stored_psycap_average = (
    df.loc[
        sample_index,
        "PsyCap_avg"
    ]
)

print(
    "Manual PsyCap average:",
    manual_psycap_average
)

print(
    "Stored PsyCap average:",
    stored_psycap_average
)

print(
    "Difference:",
    abs(
        manual_psycap_average
        -
        stored_psycap_average
    )
)
```

Check Individual Construct Calculation

```python
sample_hope_manual = (
    df.loc[
        sample_index,
        [
            "PsyCap1",
            "PsyCap2",
            "PsyCap3"
        ]
    ]
    .mean()
)

sample_hope_stored = (
    df.loc[
        sample_index,
        "Hope_avg"
    ]
)

print(
    "Manual Hope average:",
    sample_hope_manual
)

print(
    "Stored Hope average:",
    sample_hope_stored
)
```

Inspect Demographic Variables

```python
for column in DEMOGRAPHIC_COLUMNS:

    print(
        f"\n{column}"
    )

    print(
        df[column]
        .value_counts(
            dropna=False
        )
    )
```

Demographic Percentage Summary

```python
demographic_summary = []

for column in DEMOGRAPHIC_COLUMNS:

    counts = (
        df[column]
        .value_counts(
            dropna=False
        )
    )

    percentages = (
        df[column]
        .value_counts(
            normalize=True,
            dropna=False
        )
        * 100
    )

    for category in counts.index:

        demographic_summary.append({
            "Variable": column,
            "Category": category,
            "Count": int(
                counts[
                    category
                ]
            ),
            "Percentage": float(
                percentages[
                    category
                ]
            )
        })

demographic_summary = pd.DataFrame(
    demographic_summary
)

demographic_summary
```

Check data_source Distribution

```python
if "data_source" in df.columns:

    print(
        df[
            "data_source"
        ]
        .value_counts(
            dropna=False
        )
    )
```

```python
print(
    "Prepared dataset shape:",
    df.shape
)
```

Create Main Prepared Dataset

Save Demographic Summary

```python
DEMOGRAPHIC_REPORT_PATH = (
    REPORT_DIR
    / "Demographic_Summary.csv"
)

demographic_summary.to_csv(
    DEMOGRAPHIC_REPORT_PATH,
    index=False
)

print(
    "Demographic summary saved:",
    DEMOGRAPHIC_REPORT_PATH
)
```

Create Preprocessing Validation Summary

```python
preprocessing_validation = pd.DataFrame({

    "Validation_Check": [
        "Required columns",
        "Duplicate Response_ID",
        "Exact duplicate rows",
        "Missing required values",
        "Likert value range",
        "Construct value range",
        "PsyCap calculation verification"
    ],

    "Status": [
        (
            "PASS"
            if len(
                missing_columns
            ) == 0
            else "FAIL"
        ),

        (
            "PASS"
            if duplicate_response_ids == 0
            else "REVIEW"
        ),

        (
            "PASS"
            if exact_duplicate_rows == 0
            else "REVIEW"
        ),

        (
            "PASS"
            if total_missing == 0
            else "REVIEW"
        ),

        (
            "PASS"
            if len(
                invalid_likert_values
            ) == 0
            else "FAIL"
        ),

        (
            "PASS"
            if len(
                invalid_constructs
            ) == 0
            else "FAIL"
        ),

        (
            "PASS"
            if np.isclose(
                manual_psycap_average,
                stored_psycap_average
            )
            else "FAIL"
        )
    ]
})

preprocessing_validation
```

Save Validation Summary

## Step 01 Conclusion

The Phase 03 Psychological Capital → Employability dataset was successfully prepared for further statistical and machine-learning analysis.

The preprocessing stage validated the required questionnaire columns, checked duplicate responses and missing values, and confirmed that the Psychological Capital and Employability item responses were within the valid 1–5 Likert range.

Four Psychological Capital construct scores were created:

- Hope
- Optimism
- Resilience
- Self-Efficacy

An overall Psychological Capital score was also calculated.

The existing Employability structure was preserved, and an overall Employability score (`EVS_avg`) was created as the prediction target.

Separate item-level, construct-level, and overall Psychological Capital feature datasets were saved to support objective feature-representation comparison in later machine-learning stages.

The prepared dataset is now ready for Step 02 — Reliability Analysis.
