# 02 Descriptive Statistics

## Objective

The objective of this analysis is to summarize the characteristics of the collected dataset before conducting reliability analysis and machine learning. Descriptive statistics provide an overview of participant demographics and questionnaire responses, helping to understand the overall distribution of the data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path
```

```python
PROJECT_DIR = Path.cwd().parent

df = pd.read_csv(
    PROJECT_DIR /
    "data" /
    "processed" /
    "cleaned_dataset.csv"
)

df.head()
```

```python
print("="*70)
print("DATASET OVERVIEW")
print("="*70)

print("Rows :",df.shape[0])
print("Columns :",df.shape[1])

print("\nData Types\n")

print(df.dtypes)
```

```python
gender=df["Gender"].value_counts()

gender
```

```python
gender_percentage=(
    df["Gender"]
    .value_counts(normalize=True)
    *100
)

gender_percentage.round(2)
```

```python
gender.plot(kind="bar")

plt.title("Gender Distribution")

plt.xlabel("Gender")

plt.ylabel("Frequency")

plt.show()
```

```python
age=df["Age"].value_counts()

age
```

```python
(
df["Age"]
.value_counts(normalize=True)
*100
).round(2)
```

```python
age.plot(kind="bar")

plt.title("Age Distribution")

plt.show()
```

```python
degree=df["Degree_Program"].value_counts()

degree
```

```python
(
df["Degree_Program"]
.value_counts(normalize=True)
*100
).round(2)
```

```python
degree.plot(kind="bar")

plt.title("Degree Program")

plt.show()
```

```python
sie_columns=[
"SIE1",
"SIE2",
"SIE3",
"SIE4",
"SIE5",
"SIE6",
"SIE7",
"SIE8",
"SIE9",
"SIE10",
"SIE11",
"SIE12"
]
```

```python
sie_statistics=df[sie_columns].describe()

sie_statistics
```

```python
sie_statistics.T
```

```python
emp_columns=[
"EMP1",
"EMP2",
"EMP3",
"EMP4",
"EMP5",
"EMP6",
"EMP7",
"EMP8"
]
```

```python
df[emp_columns].describe().T
```

```python
pd.DataFrame({

"Mean":df[sie_columns].mean(),

"Std":df[sie_columns].std(),

"Minimum":df[sie_columns].min(),

"Maximum":df[sie_columns].max()

})
```

```python
pd.DataFrame({

"Mean":df[emp_columns].mean(),

"Std":df[emp_columns].std(),

"Minimum":df[emp_columns].min(),

"Maximum":df[emp_columns].max()

})
```

```python
df["Clear_Objectives"]=df[
["SIE1","SIE2","SIE3","SIE4"]
].mean(axis=1)

df["School_Support"]=df[
["SIE5","SIE6"]
].mean(axis=1)

df["Workplace_Environment"]=df[
["SIE7","SIE8","SIE9","SIE10"]
].mean(axis=1)

df["Career_Development"]=df[
["SIE11","SIE12"]
].mean(axis=1)

df["Employability_Score"]=df[
emp_columns
].mean(axis=1)
```

```python
constructs=[
"Clear_Objectives",
"School_Support",
"Workplace_Environment",
"Career_Development",
"Employability_Score"
]

df[constructs].describe().T
```

```python
df["Employability_Score"].hist(
bins=10
)

plt.title("Employability Score Distribution")

plt.show()
```

```python
df["Clear_Objectives"].hist(
bins=10
)

plt.title("Clear Objectives Distribution")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
```

```python
df["School_Support"].hist(
bins=10
)

plt.title("School Support Distribution")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
```

```python
df["Workplace_Environment"].hist(
bins=10
)

plt.title("Workplace Environment Distribution")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
```

```python
df["Career_Development"].hist(
bins=10
)

plt.title("Career Development Distribution")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
```

```python
df[constructs].plot(
kind="box",
figsize=(10,6)
)

plt.title("Construct Score Distribution")

plt.show()
```

```python
df.to_csv(

PROJECT_DIR/

"data"/

"processed"/

"descriptive_dataset.csv",

index=False

)
```

## Summary

Descriptive statistics were performed to summarize the characteristics of the collected data. Participant demographics including gender, age, and degree programme were analysed using frequency distributions and percentages. The questionnaire responses were summarized using the mean, standard deviation, minimum, and maximum values.

Furthermore, construct-level scores were computed by averaging the questionnaire items belonging to each construct. These construct scores provide an overall representation of Internship Experience and Employability, and they will be used in subsequent statistical analysis and machine learning model development.
