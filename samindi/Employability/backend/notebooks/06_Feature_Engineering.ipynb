# 06 Feature Engineering

## Objective

The objective of this notebook is to prepare the dataset for machine learning model development. Feature engineering transforms the processed questionnaire responses into meaningful predictive features based on the conceptual framework. This step includes selecting relevant variables, constructing the final machine learning dataset, checking feature distributions, preparing multiple feature sets, splitting the data into training and testing subsets, and scaling features where appropriate.

```python
import pandas as pd
import numpy as np

from pathlib import Path

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import StandardScaler
```

```python
PROJECT_DIR = Path.cwd().parent

df = pd.read_csv(
    PROJECT_DIR /
    "data" /
    "processed" /
    "descriptive_dataset.csv"
)

df.head()
```

```python
print(df.shape)

print(df.columns.tolist())
```

```python
features_A = [

    "Clear_Objectives",

    "School_Support",

    "Workplace_Environment",

    "Career_Development"

]

target = "Employability_Score"

X_A = df[features_A]

y = df[target]
```

```python
features_B = [

    "Gender",

    "Age",

    "Degree_Program",

    "Clear_Objectives",

    "School_Support",

    "Workplace_Environment",

    "Career_Development"

]
```

```python
X_B = df[features_B].copy()

X_B = pd.get_dummies(

    X_B,

    columns=[

        "Gender",

        "Age",

        "Degree_Program"

    ],

    drop_first=True

)
```

```python
print(X_A.head())

print(X_B.head())
```

```python
print(X_A.isnull().sum())

print(X_B.isnull().sum())
```

```python
X_A.describe()
```

```python
X_B.describe()
```

```python
X_train_A, X_test_A, y_train, y_test = train_test_split(

    X_A,

    y,

    test_size=0.20,

    random_state=42

)
```

```python
X_train_B, X_test_B, y_train_B, y_test_B = train_test_split(

    X_B,

    y,

    test_size=0.20,

    random_state=42

)
```

```python
scaler_A = StandardScaler()

X_train_A_scaled = scaler_A.fit_transform(
    X_train_A
)

X_test_A_scaled = scaler_A.transform(
    X_test_A
)
```

```python
scaler_B = StandardScaler()

X_train_B_scaled = scaler_B.fit_transform(
    X_train_B
)

X_test_B_scaled = scaler_B.transform(
    X_test_B
)
```

```python
print(X_train_A.shape)

print(X_test_A.shape)

print(y_train.shape)

print(y_test.shape)
```

```python
OUTPUT_DIR = PROJECT_DIR / "data" / "ml"

OUTPUT_DIR.mkdir(exist_ok=True)

X_train_A.to_csv(
    OUTPUT_DIR / "X_train_A.csv",
    index=False
)

X_test_A.to_csv(
    OUTPUT_DIR / "X_test_A.csv",
    index=False
)

X_train_B.to_csv(
    OUTPUT_DIR / "X_train_B.csv",
    index=False
)

X_test_B.to_csv(
    OUTPUT_DIR / "X_test_B.csv",
    index=False
)

y_train.to_csv(
    OUTPUT_DIR / "y_train.csv",
    index=False
)

y_test.to_csv(
    OUTPUT_DIR / "y_test.csv",
    index=False
)
```

## Feature Engineering

Feature engineering was performed to prepare the dataset for machine learning. Two feature sets were developed.

Model A consisted of the Internship Experience constructs derived from the conceptual framework.

Model B extended Model A by including demographic variables to evaluate whether participant characteristics improved predictive performance.

Categorical variables were encoded using one-hot encoding. The dataset was divided into training (80%) and testing (20%) subsets using a fixed random state to ensure reproducibility. Feature scaling using StandardScaler was applied where appropriate for regression-based models.
